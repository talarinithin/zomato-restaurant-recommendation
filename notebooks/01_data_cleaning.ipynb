```python
drop_cols = [
    "url",
    "phone",
    "address",
    "reviews_list",
    "menu_item"
]

df = df.drop(columns=drop_cols)
print("Remaining columns:", df.columns)
```

```python
import numpy as np

def clean_rate(rate):
    if isinstance(rate, str):
        if rate in ['NEW', '-']:
            return np.nan
        return float(rate.split('/')[0])
    return rate

df['rate'] = df['rate'].apply(clean_rate)
```

```python
def clean_cost(cost):
    if isinstance(cost, str):
        cost = cost.replace(',', '')
        return float(cost)
    return cost

df['approx_cost(for two people)'] = df['approx_cost(for two people)'].apply(clean_cost)
```

```python
df['online_order'] = df['online_order'].map({'Yes': 1, 'No': 0})
df['book_table'] = df['book_table'].map({'Yes': 1, 'No': 0})

# Fill missing values with mode
df['online_order'].fillna(df['online_order'].mode()[0], inplace=True)
df['book_table'].fillna(df['book_table'].mode()[0], inplace=True)
```

```python
non_veg_keywords = [
    'Chicken', 'Mutton', 'Fish', 'Seafood', 'Biryani',
    'Kebab', 'Prawn', 'Meat'
]

def classify_veg_nonveg(cuisine):
    if not isinstance(cuisine, str):
        return 'Unknown'
    
    cuisine_lower = cuisine.lower()
    non_veg_found = any(word.lower() in cuisine_lower for word in non_veg_keywords)
    
    if non_veg_found:
        return 'Non-Veg'
    else:
        return 'Veg'

df['veg_nonveg_type'] = df['cuisines'].apply(classify_veg_nonveg)
```

```python
df['veg_nonveg_type'].value_counts()
```

```python
veg_keywords = [
    'South Indian', 'North Indian', 'Bakery', 'Desserts',
    'Cafe', 'Fast Food', 'Vegetarian', 'Pure Veg'
]

non_veg_keywords = [
    'Chicken', 'Mutton', 'Fish', 'Seafood',
    'Biryani', 'Kebab', 'Prawn', 'Meat'
]

def classify_veg_nonveg_advanced(cuisine):
    if not isinstance(cuisine, str):
        return 'Unknown'
    
    cuisine_lower = cuisine.lower()
    
    veg_found = any(word.lower() in cuisine_lower for word in veg_keywords)
    nonveg_found = any(word.lower() in cuisine_lower for word in non_veg_keywords)
    
    if veg_found and nonveg_found:
        return 'Both'
    elif nonveg_found:
        return 'Non-Veg'
    elif veg_found:
        return 'Veg'
    else:
        return 'Veg'  # default safe assumption

df['veg_nonveg_type'] = df['cuisines'].apply(classify_veg_nonveg_advanced)
```

```python
df['veg_nonveg_type'].value_counts()
```

```python
def extract_famous_dish(row):
    # Priority 1: dish_liked
    if isinstance(row['dish_liked'], str) and row['dish_liked'].strip() != "":
        return row['dish_liked'].split(',')[0].strip()
    
    # Priority 2: cuisines
    if isinstance(row['cuisines'], str) and row['cuisines'].strip() != "":
        return row['cuisines'].split(',')[0].strip()
    
    # Fallback
    return "Not Specified"

df['famous_dish'] = df.apply(extract_famous_dish, axis=1)
```
